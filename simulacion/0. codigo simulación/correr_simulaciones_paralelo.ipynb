```python
import subprocess
import sys
import multiprocessing

def correr_paralelo_consola(seeds, script="run_sim.py"):
    max_procesos = multiprocessing.cpu_count()
    print(f"🧠 Usando hasta {max_procesos} procesos simultáneos.")

    procesos = []

    for seed in seeds:
        print(f"🚀 Ejecutando simulación con seed={seed}")
        p = subprocess.Popen([sys.executable, script, str(seed)])
        procesos.append(p)

        if len(procesos) >= max_procesos:
            for p in procesos:
                p.wait()
            procesos = []

    # Esperar procesos restantes
    for p in procesos:
        p.wait()

    print("✅ Todas las simulaciones han terminado.")
```

```python
seeds = list(range(0, 22))
correr_paralelo_consola(seeds)
```
