```python
import subprocess
import sys

def correr_paralelo_consola(seeds, script="run_sim.py"):
    procesos = []
    for seed in seeds:
        print(f"🚀 Ejecutando simulación con seed={seed}")
        p = subprocess.Popen([sys.executable, script, str(seed)])
        procesos.append(p)

    for p in procesos:
        p.wait()
    print("✅ Todas las simulaciones han terminado.")
```

```python
seeds = list(range(11, 21))
correr_paralelo_consola(seeds)
seeds = list(range(21, 31))
correr_paralelo_consola(seeds)
seeds = list(range(31, 41))
correr_paralelo_consola(seeds)
```
