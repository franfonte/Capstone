```python
dict_costo_espera_wl = {5: {
        1: 4.4378762,
        2: 4.180640861,
        3: 4.032732341
    },
    6: {
        1: 6.836340896,
        2: 5.810997239,
        3: 5.181694068
    },
    7: {
        1: 5.337300461,
        2: 4.589279284,
        3: 4.301638908
    },
    8: {
        1: 12.83250264,
        2: 10.0173047,
        3: 7.378852624
    }}

def prioridad_sacado_wl(dict_costo_espera_wl): # Revisado, funciona bien
        lista_prioridad = []
        for grd in (5, 6, 7, 8): # Solo los que estan en WL
            for requerimiento in (1, 2, 3): # Los tres requerimientos
                lista_prioridad.append((grd, requerimiento, dict_costo_espera_wl[grd][requerimiento]))
        lista_prioridad.sort(key=lambda x: x[2], reverse=True) # Quedan ordenados de mayor costo de espera a menor
        return lista_prioridad

listeilor = prioridad_sacado_wl(dict_costo_espera_wl)

for grd, requerimiento, costo in listeilor:
    print(f"GRD: {grd}, Requerimiento: {requerimiento}, Costo de espera: {costo:.2f}")
```
